# Creating ASDF files
The flexibility of the ASDF format makes it useful for saving a variety of data. This tutorial will cover the basic process of creating a new ASDF file using the python library [asdf](https://github.com/asdf-format/asdf).

```python
!pip install asdf
```

## AsdfFile

Creating an ASDF file starts with creating an [AsdfFile](https://asdf.readthedocs.io/en/latest/api/asdf.AsdfFile.html).

```python
import asdf
af = asdf.AsdfFile()
```

# Dictionary-like interface
[AsdfFile](https://asdf.readthedocs.io/en/latest/api/asdf.AsdfFile.html) instances can be used like a dictionary to store key value pairs.

```python
af["my_key"] = "my_value"
af["some_integer"] = 1
af["some_float"] = 3.14
af.info()
```

# Hierarchical structure
Values can be nested (lists inside lists, dicts inside dicts, lists inside dicts, etc).

af["my_nested_structure"] = {
    "one_deep": {
        "two_deep": [1, 2, 3]
    }
}
af.info()

# Custom objects
ASDF files can contain any value that asdf can serialize (not just basic numerical and string types). In this example we'll add a [numpy](https://numpy.org/) array (that will be saved as binary data to the ASDF file).

Developers can write asdf [extensions](https://asdf.readthedocs.io/en/latest/asdf/extending/extensions.html) to support any custom type. For example, many [Astropy](https://www.astropy.org/) objects can be saved to ASDF if the [asdf-astropy](https://github.com/astropy/asdf-astropy) plugin is installed (using `pip install asdf-astropy`).

```python
import numpy as np
af["my_array"] = np.zeros((2048, 2048), dtype=np.uint8)
af.info()
```

# Writing to disk
So far we've constructed an in-memory ASDF tree that we now can save to disk using
[AsdfFile.write_to](https://asdf.readthedocs.io/en/latest/api/asdf.AsdfFile.html#asdf.AsdfFile.write_to).

```python
af.write_to("my_file.asdf")
```

# Opening our file
Finally let's open the file with [asdf.open](https://asdf.readthedocs.io/en/latest/api/asdf.open.html) to check the contents were written to disk.

```python
read_af = asdf.open("my_file.asdf")
read_af.info()
```
